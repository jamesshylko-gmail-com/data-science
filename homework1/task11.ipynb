### 11. Создайте кортеж из 6 элементов разных типов данных и выведите на экран только те элементы, тип которых является числовым (int или float). 

P.S. Сложность: временная O(1), по памяти - O(1)

```python
mixed_numbers = (1, "seven",  ["red", "green", "blue"], 12.0, False, None)

print(*(item for item in mixed_numbers if type(item) in (int, float)))
```
