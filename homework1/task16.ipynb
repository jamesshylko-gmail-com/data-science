### 16. Создайте список чисел от 1 до 20, а затем создайте новый список, содержащий числа, которые делятся и на 2, и на 3 одновременно. 

P.S. Сложность: временная O(n), по памяти - O(n)

```python
base_list = list(range(1, 21))

updated_list = [item for item in base_list if item % 6 == 0]

print(updated_list)
```
