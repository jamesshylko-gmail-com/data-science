### 9. Напишите функцию, которая принимает строку и возвращает True, если строка является палиндромом (игнорируя регистр и пробелы), и False в противном случае. 

P.S. Сложность: временная O(m*n) m - количество элементов в тестовом кортеже, по памяти - O(m*n)

```python
def is_palindrom(user_string):
    if not user_string:
        return True
    trimmed_urer_string = lambda s: s.replace(" ", "").lower()
    return trimmed_urer_string(user_string) == trimmed_urer_string(user_string)[::-1]

palindrom_candidates = (
    "Hello, world!",
    "Аргентина манит негра",
    "шалаш",
    " Шала ш",
    "Rats live on no evil star",
    "",
    None
)

print(*(is_palindrom(item) for item in palindrom_candidates))

```
