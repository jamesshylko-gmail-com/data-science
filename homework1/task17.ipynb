### 17. Напишите программу, которая принимает от пользователя строку и выводит количество гласных букв в ней. 

P.S. Сложность: временная O(n*m), по памяти - O(1)

```python
VOWELS = "aeiouyаеёиоуыэюя"

user_string = input("Введите произвольную строку на русском или английском языке : ")

print(f"{user_string} : {sum(1 for char in user_string if char in VOWELS)}")
```
