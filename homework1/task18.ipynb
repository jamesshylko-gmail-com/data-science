### 18. Создайте кортеж, содержащий 4 элемента различных типов данных, и распакуйте первые три элемента в отдельные переменные. 

P.S. Сложность: временная O(1), по памяти - O(1)

```python
remark_info = (
    "Эрих Мария Ремарк",
    1898, 
    1970,
    {
        "На Западном фронте без перемен" : 1500000,
        "Три товарища" : 1000000,
        "Триумфальная арка" : 500000
    }
)

author, birth_year, death_year, *_ = remark_info
print(f"{author} ({birth_year}-{death_year})")
```
