### 7. Напишите программу, которая принимает от пользователя целое число и выводит таблицу умножения для этого числа от 1 до 10. 

P.S. Сложность: временная O(n), по памяти - O(1)

```python
number = input("Вседите целое число : ")

try:
    verified_number = int(number)
    print(*(f"{verified_number} \u00D7 {i} = {verified_number * i}\n" for i in range(1, 11)))
except ValueError:
    print("Error: That is not a valid integer.")
```
