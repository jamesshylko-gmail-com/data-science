### 3. Create a dict with student info: name, age, year, discinlines. Output all student discinline comma separated.

P.S. Полагаю, что раз в коде присутствует полный проход по списку предметов, то временная сложность всего алгоритма O(n). Наличие списка также, на мой взгляд, определяет сложность по памяти O(n)

```python
import textwrap

student = {
    "name": "Christopher Harington", 
    "age": 40, 
    "year": 3,
    "discinlines": [
        "Introduction to Artificial Intelligence",
        "Business Communication with Artificial Intelligence",
        "Computer Concepts",
        "Operating Systems",
        "Networking",
        "Microsoft Office with Copilot"
    ]
}

discinlines = ", ".join(student.get("discinlines"))

print(textwrap.fill(discinlines, width=120))
```
