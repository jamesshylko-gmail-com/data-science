### 15. Напишите функцию, которая принимает список чисел и возвращает кортеж из двух элементов: минимального и максимального числа. 

P.S. Сложность: временная O(1), по памяти - O(1) (алгоритм не создает дополнительных объемных структур). Насколько я понял, то если бы на вход приходил кортеж, то сложность изменилась бы на O(n)

```python
def min_max(list):
    return {min(list), max(list)}

print(min_max([12, 59, 68.0, -48.102, 26, 457888888888, -1.1e+1]))
```
