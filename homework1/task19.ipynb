### 19. Напишите функцию, которая принимает список словарей с информацией о студентах и выводит информацию о всех студентах в формате "Имя: [имя], Возраст: [возраст], Курс: [курс]". 

P.S. Сложность: временная O(n), по памяти - O(1)

```python
def print_student_info(users):
    for student in users:
        print(f"Имя: {student.get("name")}, Возраст: {student.get("age")}, Курс: {student.get("year")}")

history_faculty = [
    {
        "name" : "Tom Tompson",
        "age" : 19,
        "year" : 1
    },
    {
        "name" : "Vudro Wilson",
        "age" : 20,
        "year" : 2
    },
    {
        "name" : "Jessica Fernandez",
        "age" : 19,
        "year" : 1
    },
    {
        "name" : "Marchello Mastroiani jr.",
        "age" : 23,
        "year" : 4
    },
    {
        "name" : "Sean Connery",
        "age" : 24,
        "year" : 5
    }
]

print_student_info(history_faculty)
```
