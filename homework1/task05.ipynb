### 5. Напишите функцию, которая принимает список чисел и возвращает их среднее арифметическое. 

P.S. Сложность: временная O(n), по памяти - O(1)

```python
def custome_average(numbers):
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)

print(custome_average([15, 869.2, 346, 786, 42.26]))
```
