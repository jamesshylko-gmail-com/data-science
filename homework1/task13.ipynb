### 13. Создайте словарь, содержащий информацию о книге (автор, название, год издания, жанр), и выведите информацию в формате: "Название: [название], Автор: [автор], Год: [год], Жанр: [жанр]". 

P.S. Сложность: временная O(1), по памяти - O(1)

```python
import textwrap

book = {
    "автор" : "Франца Кафка", 
    "название" : "Превращение", 
    "год издания" : 1916, 
    "жанр" : "новелла, написанная в русле модернизма, абсурда и экзистенциализма с элементами философской притчи и сюрреализма"
}

print(textwrap.fill(f"Название: {book.get("название")}, Автор: {book.get("автор")}, Год: {book.get("год издания")}, Жанр: {book.get("жанр")}", width=120))
```
