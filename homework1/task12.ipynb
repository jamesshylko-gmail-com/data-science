### 12. Напишите программу, которая принимает от пользователя два числа и выводит результат их деления с обработкой исключения деления на ноль. 

P.S. Сложность: временная O(1), по памяти - O(1)

```python
numenator, denomenator = input("Enter two number separated by space: ").split()

try:
    print(f"{numenator} / {denomenator} = {float(numenator)/float(denomenator)}")
except ZeroDivisionError:
    print(F"{numenator} / {denomenator} : divizion by zero is forbidden")
```
