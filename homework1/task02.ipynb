### 2. Напишите программу, которая принимает строку от пользователя и выводит количество символов в этой строке без учета пробелов. 

P.S. Сложность: временная O(n), по памяти - O(n)

```python
input_string = input("Enter any string: ")

input_string_without_spaces = input_string.replace(" ", "")

string_size = len(input_string_without_spaces)

print(f"'{input_string}' > '{input_string_without_spaces}' : {string_size}")
```
