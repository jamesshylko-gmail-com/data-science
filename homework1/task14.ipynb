### 14. Создайте два множества с названиями городов, найдите города, которые есть в первом множестве, но отсутствуют во втором. 

P.S. Сложность: временная O(n + m), по памяти - O(n)

```python
european_capitals = {
    "Париж",
    "Рейкьявик",
    "Берлин",
    "Минск",
    "Валлетта"
}

million_plus_cities = {
    "Париж",
    "Токио",
    "Киншасса",
    "Монреаль",
    "Минск",
    "Рио-де-Жанейро"
}

print(european_capitals - million_plus_cities)
```
