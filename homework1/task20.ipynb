### 20. Создайте список и добавьте в него 5 различных элементов разных типов данных. Выведите на экран тип данных каждого элемента списка 

P.S. Сложность: временная O(1), по памяти - O(1)

```python
import textwrap

list = []

list.append(12.5)
list.append(None)
list.append(i for i in range(10))
list.append(textwrap.fill)
list.append(range(5))

print([type(item).__name__ for item in list])
```
