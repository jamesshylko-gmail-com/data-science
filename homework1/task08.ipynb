### 8. Создайте словарь с пятью парами «ключ-значение», где ключи - названия фруктов, а значения - их цвета. Выведите только те фрукты, цвет которых содержит букву 'к'. 

P.S. Сложность: временная O(n) (n - количество пар), по памяти - O(1)

```python
fruits = {
    "дыня" : "желтый",
    "арбуз" : "зеленый",
    "клубника" : "красный",
    "саподилла " : "коричневый",
    "слива" : "синий"
}

print(*(item[0] for item in fruits.items() if "к" in item[1]))
```
