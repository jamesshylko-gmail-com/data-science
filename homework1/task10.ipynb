### 10. Создайте список из 5 строк и замените в нем каждый третий символ в каждой строке на звездочку '*'. 

P.S. Сложность: временная O(m*n), по памяти - O(1)

```python
user_list = [
    "Carpe diem",
    "Veni, vidi, vici",
    "Status quo",
    "********",
    "Ou"
]

def third_char_changer(text):
    if not text or len(text) < 3:
        return text
    return text[:2] + "*" + text[3:]

print(*(third_char_changer(item) + "\n" for item in user_list))
```
