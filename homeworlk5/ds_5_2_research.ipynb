2) Выведем сводную таблицу, отсортированную по R², и выберем лучших 5 алгоритмов для дальнейшего исследования

```python
import warnings

from scipy.linalg import LinAlgWarning

from private.utils import load_reduced_bitcoin_data, memory_check
from public.models import (
    ADA_BST,
    BAYES,
    CAT_BST,
    DEC_TR,
    EL_NET,
    G_BOOST,
    HUBER,
    KNN,
    LASSO,
    LGBM,
    LIN_REG,
    PASS_AGR,
    RAN_FOR,
    RIDGE,
    XBT_SL,
    XGB,
    RegressionProcessor,
)

with memory_check(), warnings.catch_warnings():
    warnings.simplefilter("ignore", category=LinAlgWarning)

    df = load_reduced_bitcoin_data()
    processor = RegressionProcessor(df, "target_return")   
    config = {
        LIN_REG : {},
        RAN_FOR : {},
        LGBM : {},
        G_BOOST : {},
        XGB : {},
        XBT_SL : {},
        CAT_BST : {},
        ADA_BST : {},
        LASSO : {},
        RIDGE : {},
        BAYES : {},
        HUBER : {},
        DEC_TR : {},
        PASS_AGR : {},
        KNN : {},
        EL_NET : {}
    }
    processor.calculate(config)
    processor.summary_table(config.keys())
```
