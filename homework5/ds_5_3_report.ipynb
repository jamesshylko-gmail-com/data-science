#### Выводы:
1) Для временных рядов мы не можем доверять перемешанным данным
2) Для неперемешанных данных, очевидно, получили слишком недообученные модели
3) Формально мне не сложно провести оптимизацию Optuna для любого из алгоритмов, но, думаю, для неподходящей (недообученной) модели в этом нет никакого смысла
4) Полагаю, что главная проблема недообученных моделей в недостаточном освоении доменной области (биткойны). Возможно, стоит выбрать другие признаки или преобразовать признаки
